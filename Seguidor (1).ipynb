```python
import time
import math
from coppeliasim_zmqremoteapi_client import RemoteAPIClient

print("✅ Librerías importadas correctamente")
```

```python
print("Conectando a CoppeliaSim...")

# Conectar al servidor
client = RemoteAPIClient()
sim = client.getObject('sim')

print("✅ Conectado a CoppeliaSim")
```

```python
print("🔍 Buscando objetos en la escena...")
print("="*60)

# Obtener handles de los joints (motores)
left_joint = sim.getObject('LeftJoint')
right_joint = sim.getObject('RightJoint')

# Obtener handles de los sensores
left_sensor = sim.getObject('LeftSensor')
middle_sensor = sim.getObject('MiddleSensor')
right_sensor = sim.getObject('RightSensor')

# Obtener handle del robot
robot = sim.getObject('LineTracer')

# Mostrar resultados
print("\n📊 VERIFICACIÓN DE OBJETOS:")
print("-"*60)
print(f"LeftJoint (motor izquierdo):   {'✅ OK' if left_joint >= 0 else '❌ ERROR'} (handle: {left_joint})")
print(f"RightJoint (motor derecho):    {'✅ OK' if right_joint >= 0 else '❌ ERROR'} (handle: {right_joint})")
print(f"LeftSensor:                    {'✅ OK' if left_sensor >= 0 else '❌ ERROR'} (handle: {left_sensor})")
print(f"MiddleSensor:                  {'✅ OK' if middle_sensor >= 0 else '❌ ERROR'} (handle: {middle_sensor})")
print(f"RightSensor:                   {'✅ OK' if right_sensor >= 0 else '❌ ERROR'} (handle: {right_sensor})")
print(f"LineTracer (robot):            {'✅ OK' if robot >= 0 else '❌ ERROR'} (handle: {robot})")
print("-"*60)

# Verificar que todos los objetos existen
if left_joint < 0 or right_joint < 0:
    print("\n❌ ERROR CRÍTICO: No se encontraron los motores")
    print("Verifica que los nombres sean exactamente 'LeftJoint' y 'RightJoint'")
    
if left_sensor < 0 or middle_sensor < 0 or right_sensor < 0:
    print("\n⚠️  ADVERTENCIA: Faltan algunos sensores")
    print("Verifica que los sensores se llamen: LeftSensor, MiddleSensor, RightSensor")
    
print("\n✅ Configuración de handles completada")
```

```python
print("🔧 CONFIGURANDO ROBOT SEGUIDOR DE LÍNEA")
print("="*60)

# Verificar conexión
try:
    # Probar conexión simple
    print("✅ Conexión con CoppeliaSim activa")
except:
    print("❌ ERROR: No hay conexión con CoppeliaSim")
    raise

# Función para buscar objetos por nombre
def buscar_objeto(nombre):
    """Busca un objeto en la escena"""
    try:
        handle = sim.getObject(nombre)
        if handle >= 0:
            return handle
    except:
        pass
    
    try:
        handle = sim.getObject('./' + nombre)
        if handle >= 0:
            return handle
    except:
        pass
    
    return -1

# Buscar motores con los nombres de tu escena
print("\n🔍 BUSCANDO OBJETOS...")
print("-"*60)

# Motores
motor_izq = buscar_objeto('LeftJoint')
motor_der = buscar_objeto('RightJoint')

# Si no encuentra LeftJoint/RightJoint, buscar alternativas
if motor_izq < 0:
    motor_izq = buscar_objeto('DynamicLeftJoint')
    motor_der = buscar_objeto('DynamicRightJoint')
    print("Usando nombres alternativos: DynamicLeftJoint/DynamicRightJoint")

# Sensores
sensor_izq = buscar_objeto('LeftSensor')
sensor_centro = buscar_objeto('MiddleSensor')
sensor_der = buscar_objeto('RightSensor')

# Robot
robot = buscar_objeto('LineTracer')
if robot < 0:
    robot = buscar_objeto('LineTracerBase')
if robot < 0:
    robot = buscar_objeto('Body')

# Mostrar resultados
print("\n📊 RESULTADOS DE BÚSQUEDA:")
print("-"*60)
print(f"Motor izquierdo (LeftJoint):    {'✅ OK' if motor_izq >= 0 else '❌ NO'} (handle: {motor_izq})")
print(f"Motor derecho (RightJoint):     {'✅ OK' if motor_der >= 0 else '❌ NO'} (handle: {motor_der})")
print(f"Sensor izquierdo (LeftSensor):  {'✅ OK' if sensor_izq >= 0 else '❌ NO'} (handle: {sensor_izq})")
print(f"Sensor centro (MiddleSensor):   {'✅ OK' if sensor_centro >= 0 else '❌ NO'} (handle: {sensor_centro})")
print(f"Sensor derecho (RightSensor):   {'✅ OK' if sensor_der >= 0 else '❌ NO'} (handle: {sensor_der})")
print(f"Robot (LineTracer):             {'✅ OK' if robot >= 0 else '❌ NO'} (handle: {robot})")
print("-"*60)

# Verificar si tenemos los objetos principales
if motor_izq >= 0 and motor_der >= 0:
    print("\n✅ MOTORES ENCONTRADOS")
else:
    print("\n❌ ERROR: No se encontraron los motores")
    print("\n🔧 SOLUCIÓN:")
    print("1. En CoppeliaSim, busca en la jerarquía los nombres de los joints")
    print("2. Los joints deben ser 'LeftJoint' y 'RightJoint'")
    print("3. Si tienen otros nombres, dímelos para ajustar el código")

if sensor_izq >= 0 and sensor_centro >= 0 and sensor_der >= 0:
    print("✅ SENSORES ENCONTRADOS")
else:
    print("\n⚠️  ADVERTENCIA: Faltan sensores")
    print("Verifica los nombres: LeftSensor, MiddleSensor, RightSensor")

# Configurar motores si se encontraron
if motor_izq >= 0 and motor_der >= 0:
    print("\n⚙️  CONFIGURANDO MOTORES...")
    try:
        # Intentar configurar modo velocidad
        sim.setJointMode(motor_izq, 1)  # 1 = velocity mode
        sim.setJointMode(motor_der, 1)
        print("✅ Motores configurados en modo velocidad")
    except Exception as e:
        print(f"⚠️  Los motores ya están configurados: {e}")

print("\n" + "="*60)
print("✅ CONFIGURACIÓN COMPLETADA")

# Guardar handles para usar en otras celdas
motor_izq_handle = motor_izq
motor_der_handle = motor_der
sensor_izq_handle = sensor_izq
sensor_centro_handle = sensor_centro
sensor_der_handle = sensor_der
robot_handle = robot

print("\n💾 Handles guardados para usar en siguientes celdas")
```

```python
# Esta celda lista todos los objetos de forma segura
print("📋 LISTANDO OBJETOS DE LA ESCENA")
print("="*60)

# Método alternativo para listar objetos
try:
    # Obtener todos los objetos por nombre conocido
    nombres_a_buscar = [
        'LeftJoint', 'RightJoint', 'DynamicLeftJoint', 'DynamicRightJoint',
        'LeftSensor', 'MiddleSensor', 'RightSensor',
        'LineTracer', 'LineTracerBase', 'Body', 'LeftWheel', 'RightWheel'
    ]
    
    print("\n🔍 Buscando objetos por nombre:")
    print("-"*60)
    
    for nombre in nombres_a_buscar:
        try:
            handle = sim.getObject(nombre)
            if handle >= 0:
                print(f"✅ {nombre} -> handle: {handle}")
            else:
                print(f"❌ {nombre} -> No encontrado")
        except:
            print(f"⚠️  {nombre} -> Error al buscar")
            
except Exception as e:
    print(f"Error: {e}")

print("-"*60)
print("\n💡 Si los nombres no coinciden, dime los nombres exactos de:")
print("   - Los joints (motores)")
print("   - Los sensores")
print("   - El robot base")
```

```python
print("🧪 PROBANDO MOVIMIENTO DE MOTORES")
print("="*60)

# Usar los handles guardados
try:
    motor_izq = motor_izq_handle
    motor_der = motor_der_handle
except:
    print("❌ Handles no definidos. Ejecuta primero la Celda 4")

if motor_izq >= 0 and motor_der >= 0:
    # Iniciar simulación
    print("\n🎬 Iniciando simulación...")
    sim.startSimulation()
    time.sleep(2)
    
    print("\n🔧 Probando motores...")
    
    try:
        # Probar ambos motores
        print("▶️  Avanzando recto...")
        sim.setJointTargetVelocity(motor_izq, 2.0)
        sim.setJointTargetVelocity(motor_der, 2.0)
        time.sleep(2)
        
        print("⏸️  Deteniendo...")
        sim.setJointTargetVelocity(motor_izq, 0)
        sim.setJointTargetVelocity(motor_der, 0)
        time.sleep(1)
        
        print("🔄 Girando derecha...")
        sim.setJointTargetVelocity(motor_izq, 2.0)
        sim.setJointTargetVelocity(motor_der, -2.0)
        time.sleep(2)
        
        print("⏸️  Deteniendo...")
        sim.setJointTargetVelocity(motor_izq, 0)
        sim.setJointTargetVelocity(motor_der, 0)
        
        print("\n✅ Prueba completada - ¿Viste movimiento en las ruedas?")
        
    except Exception as e:
        print(f"❌ Error durante prueba: {e}")
        print("\nPosibles causas:")
        print("1. Los objetos no son joints")
        print("2. Los joints no están en modo velocidad")
        print("3. Los nombres de los handles son incorrectos")
    
    # Detener simulación
    sim.stopSimulation()
    print("⏸️  Simulación detenida")
    
else:
    print(f"❌ Handles inválidos: motor_izq={motor_izq}, motor_der={motor_der}")
```

```python
print("🧪 PROBANDO SENSORES DE VISIÓN")
print("="*60)

# Usar los handles guardados
try:
    sensor_izq = sensor_izq_handle
    sensor_centro = sensor_centro_handle
    sensor_der = sensor_der_handle
    print(f"Handles: Izq={sensor_izq}, Centro={sensor_centro}, Der={sensor_der}")
except:
    print("⚠️  Handles no definidos. Ejecuta primero la Celda 4")

if sensor_izq >= 0 and sensor_centro >= 0 and sensor_der >= 0:
    # Iniciar simulación
    print("\n🎬 Iniciando simulación...")
    sim.startSimulation()
    time.sleep(2)
    
    print("\n📡 Leyendo sensores de visión...")
    print("Coloca el robot sobre la línea negra para probar")
    print("-"*60)
    
    import numpy as np
    
    def analizar_imagen(img, resolution):
        """Analiza la imagen del sensor para detectar línea negra"""
        try:
            # Convertir a array numpy
            img_array = np.array(img, dtype=np.uint8).reshape(resolution[1], resolution[0], 3)
            
            # Calcular brillo promedio (0=negro, 255=blanco)
            brillo_promedio = np.mean(img_array)
            
            # Umbral para detectar línea negra (ajustable)
            # Si el brillo es bajo (< 100), es línea negra
            detecta_linea = brillo_promedio < 100
            
            return detecta_linea, brillo_promedio
        except:
            return False, 255
    
    for i in range(15):
        # Leer sensores de visión
        img_izq, res_izq = sim.getVisionSensorImage(sensor_izq)
        img_centro, res_centro = sim.getVisionSensorImage(sensor_centro)
        img_der, res_der = sim.getVisionSensorImage(sensor_der)
        
        # Analizar cada sensor
        det_izq, brillo_izq = analizar_imagen(img_izq, res_izq)
        det_centro, brillo_centro = analizar_imagen(img_centro, res_centro)
        det_der, brillo_der = analizar_imagen(img_der, res_der)
        
        # Crear visualización
        estado = "⬛" if det_izq else "⬜"
        estado += "⬛" if det_centro else "⬜"
        estado += "⬛" if det_der else "⬜"
        
        print(f"Ciclo {i+1:2d}: [{estado}] | Brillo: L:{brillo_izq:.0f} C:{brillo_centro:.0f} R:{brillo_der:.0f}")
        time.sleep(0.5)
    
    print("\n✅ Prueba de sensores completada")
    
    # Detener simulación
    sim.stopSimulation()
    print("⏸️  Simulación detenida")
    
else:
    print("❌ No se pueden probar sensores: handles no válidos")
```

```python
print("📝 FUNCIONES DE CONTROL PARA SENSORES DE VISIÓN")
print("="*60)

import numpy as np

# Parámetros de control
velocidad_base = 2.0  # Velocidad normal (1-3)
kp = 1.2  # Ganancia de corrección (0.5-2)
umbral_linea = 100  # Umbral para detectar línea negra (menor = más sensible)

print(f"⚙️  Parámetros configurados:")
print(f"   Velocidad base: {velocidad_base}")
print(f"   Ganancia KP: {kp}")
print(f"   Umbral línea: {umbral_linea}")

def analizar_imagen_sensor(sensor_handle):
    """Analiza la imagen de un sensor de visión y retorna si detecta línea negra"""
    try:
        # Obtener imagen del sensor
        img, resolution = sim.getVisionSensorImage(sensor_handle)
        
        if img and len(img) > 0:
            # Convertir a array numpy
            img_array = np.array(img, dtype=np.uint8).reshape(resolution[1], resolution[0], 3)
            
            # Calcular brillo promedio (0=negro, 255=blanco)
            brillo_promedio = np.mean(img_array)
            
            # Detectar línea negra (brillo bajo)
            detecta_linea = brillo_promedio < umbral_linea
            
            return detecta_linea, brillo_promedio
        else:
            return False, 255
            
    except Exception as e:
        return False, 255

def leer_sensores():
    """Lee los 3 sensores de visión"""
    try:
        det_izq, brillo_izq = analizar_imagen_sensor(sensor_izq_handle)
        det_centro, brillo_centro = analizar_imagen_sensor(sensor_centro_handle)
        det_der, brillo_der = analizar_imagen_sensor(sensor_der_handle)
        
        return det_izq, det_centro, det_der, brillo_izq, brillo_centro, brillo_der
        
    except Exception as e:
        print(f"Error leyendo sensores: {e}")
        return False, False, False, 255, 255, 255

def calcular_velocidades(s_left, s_middle, s_right):
    """
    Calcula velocidades de los motores según los sensores
    Retorna: (velocidad_izquierda, velocidad_derecha, estado)
    """
    # Lógica de control para seguidor de línea
    if s_middle:  # Línea en el centro -> seguir recto
        correccion = 0
        estado = "Recto ➡️"
        
    elif s_left:   # Línea a la izquierda -> girar izquierda
        correccion = -kp
        estado = "Girar izquierda ⬅️"
        
    elif s_right:  # Línea a la derecha -> girar derecha
        correccion = kp
        estado = "Girar derecha ➡️"
        
    else:  # Perdió la línea -> girar para buscar
        correccion = kp * 1.5
        estado = "Buscando línea 🔍"
    
    # Calcular velocidades
    v_left = velocidad_base - correccion
    v_right = velocidad_base + correccion
    
    # Limitar velocidades máximas
    v_left = max(-3.0, min(3.0, v_left))
    v_right = max(-3.0, min(3.0, v_right))
    
    return v_left, v_right, estado

print("\n✅ Funciones definidas correctamente")
```

```python
print("🚗 INICIANDO SEGUIDOR DE LÍNEA CON SENSORES DE VISIÓN")
print("="*60)
print("📋 VERIFICACIÓN FINAL:")
print(f"   Motor izquierdo: {'✅' if motor_izq_handle >= 0 else '❌'}")
print(f"   Motor derecho: {'✅' if motor_der_handle >= 0 else '❌'}")
print(f"   Sensor izquierdo: {'✅' if sensor_izq_handle >= 0 else '❌'}")
print(f"   Sensor centro: {'✅' if sensor_centro_handle >= 0 else '❌'}")
print(f"   Sensor derecho: {'✅' if sensor_der_handle >= 0 else '❌'}")
print("="*60)
print("⚠️  IMPORTANTE:")
print("1. Asegúrate que CoppeliaSim está en PLAY")
print("2. El robot debe estar sobre la línea negra")
print("3. Los sensores deben apuntar al suelo")
print("4. La línea debe ser negra sobre fondo claro")
print("5. Presiona 'Stop' en Jupyter para detener")
print("="*60)
print("🎬 INICIANDO CONTROL...")
print()

# Iniciar simulación
sim.startSimulation()
time.sleep(2)

# Variables de control
ciclo = 0

try:
    while True:
        # 1. Leer sensores
        s_left, s_middle, s_right, b_left, b_middle, b_right = leer_sensores()
        
        # 2. Calcular velocidades
        v_left, v_right, estado = calcular_velocidades(s_left, s_middle, s_right)
        
        # 3. Crear visualización de sensores
        visual = "⬛" if s_left else "⬜"
        visual += "⬛" if s_middle else "⬜"
        visual += "⬛" if s_right else "⬜"
        
        # 4. Mostrar información en tiempo real
        print(f"Ciclo {ciclo:3d} | [{visual}] | {estado:20} | "
              f"Vel: ({v_left:5.2f}, {v_right:5.2f}) | "
              f"Brillo: {b_left:.0f}/{b_middle:.0f}/{b_right:.0f}")
        
        # 5. Aplicar velocidades a los motores
        sim.setJointTargetVelocity(motor_izq_handle, v_left)
        sim.setJointTargetVelocity(motor_der_handle, v_right)
        
        # 6. Esperar para el siguiente ciclo
        time.sleep(0.05)  # 50ms = 20 ciclos por segundo
        ciclo += 1
        
except KeyboardInterrupt:
    print("\n\n🛑 Usuario detuvo el programa")
    
except Exception as e:
    print(f"\n❌ Error durante ejecución: {e}")
    
finally:
    # Detener robot
    print("\n🛑 Deteniendo robot...")
    sim.setJointTargetVelocity(motor_izq_handle, 0)
    sim.setJointTargetVelocity(motor_der_handle, 0)
    time.sleep(0.5)
    
    # Detener simulación
    sim.stopSimulation()
    print("✅ Simulación detenida")
    print(f"📊 Total de ciclos ejecutados: {ciclo}")
```
